This notebook collects all flagged IPs from Notebooks 02 and 03 and sends
a `logger`-based rsyslog alarm message to the Wazuh manager for each one.

The message format expected by the custom decoder `ueba_alarm` is:
```
Alarm UEBA <ip_address>
```
Wazuh rule **100201** (level 7) fires on every such message.

## 1 — Configuration

````python
# ============================================================
# NOTEBOOK 04 — CELL 1
# SIEM endpoint configuration
# Change these values to match your deployment.
# ============================================================

SIEM_HOST = "172.100.0.12"   # Wazuh manager IP
SIEM_PORT = 514              # UDP syslog port

print(f"[config] SIEM endpoint: {SIEM_HOST}:{SIEM_PORT}")
````

## 2 — Collect flagged IPs

Paste the IPs identified by each rule in the lists below.
If a rule flagged no devices, leave the list empty.

````python
# ============================================================
# NOTEBOOK 04 — CELL 2
# Flagged IP addresses from Notebooks 02 and 03.
# Fill in the results after running those notebooks.
# ============================================================

# Rule i  — Internal BotNet activity  (R1)
flagged_r1_botnet = [
    "192.168.103.110",
    "192.168.103.153",
    "192.168.103.207",
]

# Rule ii — Data exfiltration via HTTPS / DNS  (R2 + R3)
flagged_r2_exfil = [
    "192.168.103.103",
    "192.168.103.114",
    "192.168.103.118",
    "192.168.103.153",
    "192.168.103.187",
    "192.168.103.32",
    "192.168.103.80",
    "192.168.103.81",
    "192.168.103.92",
    "192.168.103.99",
]

# Rule iii — C&C activity via DNS  (R4)
flagged_r3_cnc = [
    "192.168.103.195",
]

# Rule iv — Anomalous external destinations  (R5)
flagged_r4_dst = [
    "192.168.103.20",
    "192.168.103.21",
    "192.168.103.59",
]

# Rule v  — Anomalous external usage of corporate public servers  (R6)
flagged_r5_ext = [
    "188.83.74.162",
    "188.83.74.33",
    "188.83.74.51",
    "188.83.74.61",
    "188.83.74.64",
]

# Merge all flagged IPs (deduplicated)
all_flagged = list({
    ip
    for lst in [flagged_r1_botnet, flagged_r2_exfil, flagged_r3_cnc,
                flagged_r4_dst, flagged_r5_ext]
    for ip in lst
})

print(f"Total unique flagged IPs: {len(all_flagged)}")
for ip in sorted(all_flagged):
    print(f"  {ip}")
````

## 3 — Send rsyslog alarms

Two sending methods are provided:

* **Method A** — Python `socket` (no external dependencies, works anywhere)
* **Method B** — system `logger` command (requires the host `logger` binary)

Method A is the default and is fully self-contained.

````python
# ============================================================
# NOTEBOOK 04 — CELL 3  (fallback — not required by course)
# METHOD A — Send alarms via raw Python UDP socket
# Only use this if the `logger` binary is unavailable.
# ============================================================

import socket
import datetime

def send_ueba_alarm_socket(ip: str, host: str = SIEM_HOST, port: int = SIEM_PORT) -> None:
    """Fallback: send UEBA alarm via raw UDP syslog (RFC-3164)."""
    hostname = socket.gethostname()
    timestamp = datetime.datetime.now().strftime("%b %d %H:%M:%S")
    message = f"<134>{timestamp} {hostname} ueba: Alarm UEBA {ip}"
    with socket.socket(socket.AF_INET, socket.SOCK_DGRAM) as sock:
        sock.sendto(message.encode("utf-8"), (host, port))
    print(f"  [sent] {message}")


# Uncomment below ONLY if `logger` is unavailable (Method B, Cell 4):
# if not all_flagged:
#     print("[warn] No flagged IPs. Fill Cell 2 first.")
# else:
#     for ip in sorted(all_flagged):
#         send_ueba_alarm_socket(ip)

print("Method A (fallback) loaded — not executed. Run Cell 4 instead.")
````

````python
# ============================================================
# NOTEBOOK 04 — CELL 4
# METHOD B (default) — Send alarms via system `logger` command
# Matches the approach shown in the course guide:
#   logger -n 172.100.0.12 -P 514 "Alarm UEBA 10.1.1.1"
# ============================================================

import subprocess

def send_ueba_alarm_logger(ip: str, host: str = SIEM_HOST, port: int = SIEM_PORT) -> None:
    """Send a UEBA alarm using the system `logger` binary (course-prescribed method)."""
    cmd = ["logger", "-n", host, "-P", str(port), f"Alarm UEBA {ip}"]
    result = subprocess.run(cmd, capture_output=True, text=True)
    if result.returncode != 0:
        print(f"  [error] {ip}: {result.stderr.strip()}")
    else:
        print(f"  [sent] Alarm UEBA {ip}")


if not all_flagged:
    print("[warn] No flagged IPs to report. Fill in the lists in Cell 2 first.")
else:
    print(f"Sending {len(all_flagged)} UEBA alarm(s) to {SIEM_HOST}:{SIEM_PORT} via logger ...")
    for ip in sorted(all_flagged):
        send_ueba_alarm_logger(ip)
    print("Done.")
````

## 4 — Verify reception in Wazuh

After running Cell 3, confirm that the alarms arrived at the Wazuh manager.

### 4.1 Check raw archives log

```bash
docker exec wazuh.manager tail -f /var/ossec/logs/archives/archives.log | grep "Alarm UEBA"
```

### 4.2 Check JSON alerts (decoded)

```bash
docker exec wazuh.manager tail -f /var/ossec/logs/archives/archives.json \
  | python3 -c "import sys,json; [print(json.dumps(json.loads(l), indent=2))
                 for l in sys.stdin if 'Alarm UEBA' in l]"
```

### 4.3 Wazuh Dashboard

Go to **Menu → Explore → Discover** and use the DQL query:

```
rule.id: 100201
```

Each event should show:
* `rule.id` = **100201**
* `rule.level` = **7**
* `rule.description` = `UEBA Alarm triggered from <ip>`
* `data.srcip` = the flagged IP address

````python
# ============================================================
# NOTEBOOK 04 — CELL 5
# Summary table: rule → flagged IPs
# ============================================================

import pandas as pd

summary_rows = []
for rule_id, label, ips in [
    ("i",   "Internal BotNet activity",                      flagged_r1_botnet),
    ("ii",  "Data exfiltration via HTTPS/DNS",               flagged_r2_exfil),
    ("iii", "C&C activity via DNS",                          flagged_r3_cnc),
    ("iv",  "Anomalous external destinations",               flagged_r4_dst),
    ("v",   "Anomalous external usage of corporate servers", flagged_r5_ext),
]:
    summary_rows.append({
        "Rule": rule_id,
        "Description": label,
        "# Flagged": len(ips),
        "IP Addresses": ", ".join(sorted(ips)) if ips else "—",
    })

df_summary = pd.DataFrame(summary_rows)
df_summary.set_index("Rule", inplace=True)
print(df_summary.to_string())
````
